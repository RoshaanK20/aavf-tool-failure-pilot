## Upload the public-release archive and locate the repository


```python
from google.colab import files
from pathlib import Path
from zipfile import ZipFile
import subprocess, sys
uploaded=files.upload()
archives=[x for x in uploaded if x.lower().endswith('.zip')]
assert archives, 'Upload the AAVF public release ZIP'
extract_to=Path('/content/aavf_llm_release');extract_to.mkdir(exist_ok=True)
with ZipFile(archives[0]) as z:z.extractall(extract_to)
matches=list(extract_to.rglob('run_llm_experiment.py'))
assert len(matches)==1, matches
repo=matches[0].parent
print('Repository:',repo)
```

## Run offline tests (these use a mock backend, not an LLM)


```python
subprocess.run([sys.executable,'-m','unittest','discover','-s','tests','-p','test_*.py','-v'],cwd=repo,check=True)
```

## Optional live inference (requires model download and sufficient resources)
A three-scenario smoke test verifies that the model can select tool actions. These results are exploratory, not publishable benchmark scores.


```python
subprocess.run([sys.executable,'run_llm_experiment.py','--n','3','--model','Qwen/Qwen2.5-1.5B-Instruct'],cwd=repo,check=True)
```
