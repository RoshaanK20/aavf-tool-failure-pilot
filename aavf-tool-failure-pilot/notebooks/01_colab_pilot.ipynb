## Upload and extract the release ZIP
Upload `AAVF_Public_Release_v1.zip`. This notebook searches for the code folder automatically.


```python
from google.colab import files
from pathlib import Path
from zipfile import ZipFile
import sys, subprocess
uploaded = files.upload()
archives = [x for x in uploaded if x.lower().endswith('.zip')]
assert archives, 'Please upload a ZIP archive containing the research repository.'
extract_to = Path('/content/aavf_release')
extract_to.mkdir(exist_ok=True)
with ZipFile(archives[0]) as z:
    z.extractall(extract_to)
matching = list(extract_to.rglob('run_experiment.py'))
assert len(matching)==1, f'Expected one repo, found {len(matching)}'
repo = matching[0].parent
print('Repository:', repo)
```

## Run automated tests


```python
subprocess.run([sys.executable, '-m', 'unittest', 'discover', '-s', 'tests', '-p', 'test_*.py', '-v'],cwd=repo,check=True)
```

## Re-run all 1,600 scenarios and inspect results


```python
subprocess.run([sys.executable, 'run_experiment.py'],cwd=repo,check=True)
```

```python
import json
from IPython.display import Image, display
summary=json.loads((repo/'results'/'summary.json').read_text())
print(json.dumps(summary['policies'],indent=2))
for png in ['simulation_success.png','simulation_calls.png']:
    if (repo/'figures'/png).exists(): display(Image(filename=str(repo/'figures'/png)))
```

**Interpretation:** These are outcomes from a deterministic toy simulation. They do not establish reliability of autonomous LLM agents. The verify-all baseline assumes a perfect and instantaneous state audit.
